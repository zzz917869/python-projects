```python
import pandas as pd
data = {
    '姓名':['张三','李四','王五','赵路','小明',],
    '数学':[85,92,78,88,95],
    '英语':[90,88,85,92,80],
    '物理':[75,80,88,85,90]

}

scores = pd.DataFrame(data)
scores
```

```python
# 计算学生的总分和平均分
scores['总分'] = scores[['数学','英语','物理']].sum(axis=1)
scores
scores ['平均分'] = scores ['总分']/3
scores
scores ['平均分2']= scores[['数学','英语','物理']].mean(axis=1)
scores


```

```python
# 找出数学成绩高于90分或者英语成绩高于85分的学生
scores[(scores['数学']>90 )&(scores['英语']>85)]
r1 = scores.sort_values('总分',ascending=False).head(3)



```

```python
r2 = scores.nlargest(3,columns=['总分'] )
print(r1)
print(r2)
```

```python
import pandas as pd

data = {
    '产品单价':['A','B','C','D',],
    '单价':[100,150,200,120],
    '销量':[50,30,20,40]
}

df = pd.DataFrame(data)
df['总销售额'] = df['单价']*df['销量']
df
```

```python
df.nlargest(1,columns= ['总销售额'])
df.sort_values('总销售额',ascending=False)
```
