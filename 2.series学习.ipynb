```python
#案例
import pandas as pd
import numpy as np
np.random.seed(42)
values = np.random.randint(50,101,10)#50到101随机生成10个数字
indexes = []
for i in range(1,11):
    indexes.append('学生'+str(i))

scores = pd.Series (values,indexes)
print(scores)
print("平均分",scores.mean())
print('最高分',scores.max())
print('最低分',scores.min())
print('中位数',scores.median() )
print('众数',scores.mode() )
print(scores.sort_values())# 排序
# 找到高于平均分的人数
mean = scores.mean()

#scores [scores > mean]
#使用len 或者 count 获取人数
#len(scores [scores > mean])
#scores [scores > mean].count()
print('高于平均分的人数',len(scores [scores > mean]))
print('高于平均分人数',scores [scores > mean].count())
```

```python
# 案例分析，温度
import pandas as pd
import numpy as np
temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'] )
n = temperatures[temperatures > 30].count()
print("超过30度的天数:",n)
print('平均温度',temperatures.mean() )
t2 = temperatures .sort_values(ascending=False)# 降序
print("从高到低排序：",t2)
```

```python
# 找出温度变化的两天
#28 31 29 32 30 27 33
# diff找差值变化
#abs 绝对值
t3 = temperatures.diff().abs()

print('温度变化最大的两天',*(t3.sort_values(ascending=False).keys()[:2].tolist()) )
```

```python
#每小时销售数据分析
import pandas as pd
import numpy as np
np.random.seed(42)
h = (pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01',periods=24,freq='h')))
# 按天重新采样每日总销售额
day_sales = h.resample('D').sum()
h.sum()


```

```python
# 计算营业时间8--12和非营业时间的销售比例
h.between_time('8:00','22:00')# 筛选一段时间内的series
```

```python
mask = (h.index.hour>=8)&((h.index.hour<=22))
#print(business_hours_sales/(day_sales-business_hours_sales))
b = h[mask]
a = h[~mask]
print(b.sum()/a.sum())
```

```python
#找到销售额最高的三个小时
print(h .nlargest(3))
```
